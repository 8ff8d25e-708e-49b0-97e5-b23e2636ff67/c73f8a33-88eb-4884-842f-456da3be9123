```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create figure and axis
fig, ax = plt.subplots(figsize=(8, 8))

# Define regions as rectangles
zones = [
    {'xy': (0,0), 'width': 130, 'height': 180, 'color': 'red', 'label': 'HIGH: STAGE 2 HYPERTENSION'},
    {'xy': (0,0), 'width': 100, 'height': 160, 'color': 'orange', 'label': 'HIGH: STAGE 1 HYPERTENSION'},
    {'xy': (0,0), 'width': 90, 'height': 140, 'color': 'yellow', 'label': 'PREHYPERTENSION'},
    {'xy': (0,0), 'width': 80, 'height': 120, 'color': 'lightgreen', 'label': 'NORMAL'},
    {'xy': (0,0), 'width': 60, 'height': 90, 'color': 'green', 'label': 'LOW'},
]

# Add rectangles
for zone in zones:
    rect = patches.Rectangle(zone['xy'], zone['width'], zone['height'], linewidth=1,
                             edgecolor='white', facecolor=zone['color'])
    ax.add_patch(rect)
    x = zone['xy'][0] + 1
    y = zone['xy'][1] + zone['height'] - 5
    ax.text(x, y, zone['label'], fontsize=10, weight='bold')

# Set labels and ticks
ax.set_xlabel("DIASTOLIC BLOOD PRESSURE", fontsize=12, weight='bold')
ax.set_ylabel("SYSTOLIC BLOOD PRESSURE", fontsize=12, weight='bold')
ax.set_title("BLOOD PRESSURE CHART", fontsize=16, color='deepskyblue', weight='bold')
ax.set_xlim(0, 130)
ax.set_ylim(0, 180)

plt.grid(False)
plt.tight_layout()
plt.show()
```
